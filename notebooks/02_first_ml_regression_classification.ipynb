> 🟢 **Level A · 必须掌握** | 阶段性目标：第一次完整跑通回归和分类。

这一节只追求一个目标：**第一次把材料数据表交给机器学习模型，并看到预测结果。**

你会完成：
- 回归：预测 CO₂ uptake 是多少；
- 分类：判断一个 COF 属于高吸附还是低吸附。

本节暂时不讲复杂调参、交叉验证或 SHAP。


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_absolute_error, r2_score, accuracy_score, ConfusionMatrixDisplay
df = pd.read_csv('https://raw.githubusercontent.com/Wanteen/COF-ML-Tutorial/main/data/cof_demo.csv')
display(df.head())
print('shape =', df.shape)
```

## 1. 回归：预测连续数值

`X` 是材料特征，`y` 是 target。这里用教学数据预测 `CO2_uptake_demo`。

> `CO2_uptake_demo` 是人工教学 target，只用于理解流程。


```python
features=['pore_A','void_fraction','density','N_fraction','O_fraction']
X=df[features]
y=df['CO2_uptake_demo']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
reg=RandomForestRegressor(n_estimators=200,random_state=42)
reg.fit(X_train,y_train)
pred=reg.predict(X_test)
print('MAE =',round(mean_absolute_error(y_test,pred),3))
print('R²  =',round(r2_score(y_test,pred),3))
plt.scatter(y_test,pred)
lims=[min(y_test.min(),pred.min()),max(y_test.max(),pred.max())]
plt.plot(lims,lims,'--')
plt.xlabel('Actual'); plt.ylabel('Predicted'); plt.show()
```

记住第一次完整闭环：

`features → split → fit → predict → metric`


## 2. 分类：预测类别

为了入门，我们把同一 target 按中位数分为：
- `1 = High uptake`
- `0 = Low uptake`

科研中分类阈值必须有明确科学或工程依据；这里的中位数仅用于教学。


```python
X_train,X_test,y_train_cont,y_test_cont=train_test_split(X,df['CO2_uptake_demo'],test_size=0.25,random_state=42)
threshold=y_train_cont.median()
y_train=(y_train_cont>=threshold).astype(int)
y_test=(y_test_cont>=threshold).astype(int)
clf=RandomForestClassifier(n_estimators=200,random_state=42)
clf.fit(X_train,y_train)
pred_class=clf.predict(X_test)
print('threshold =',threshold)
print('accuracy  =',round(accuracy_score(y_test,pred_class),3))
ConfusionMatrixDisplay.from_predictions(y_test,pred_class)
plt.show()
```

## 3. Regression vs Classification

| 问题 | Regression | Classification |
|---|---|---|
| 输出 | 连续数值 | 类别 |
| COF 例子 | CO₂ uptake = 3.2 mol/kg | High / Low |
| 常见指标 | MAE、RMSE、R² | accuracy、precision、recall、F1 |
