````python
!pip -q install pymatgen
from pymatgen.core import Lattice, Structure
lattice=Lattice.from_parameters(a=18,b=22,c=3.6,alpha=90,beta=90,gamma=95)
print(lattice); print('volume =',lattice.volume)
hex_lattice=Lattice.hexagonal(a=12,c=3.5)
print('hexagonal a,b,gamma =',hex_lattice.a,hex_lattice.b,hex_lattice.gamma)
````

## 1. 一般晶胞
一般晶胞由 `a,b,c,α,β,γ` 描述。COF 不一定是六方结构，也不要求 `a=b`。


````python
s=Structure(lattice,['C','C','N','N'],[[0,0,0.5],[0.5,0.5,0.5],[0.25,0.25,0.5],[0.75,0.75,0.5]])
print(s.composition)
print('atoms =',len(s),'density =',s.density)
for site in s:
    print(site.species_string,site.frac_coords,site.coords)
````
