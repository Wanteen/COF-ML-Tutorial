## COF 背景：从分子构筑到材料性质

**共价有机框架（Covalent Organic Frameworks, COFs）** 是有机构筑单元通过共价键连接形成的晶态多孔网络。理解 COF 时，需要同时考虑“分子如何连接”和“结构如何在空间周期性重复”。2005 年报道的 COF-1、COF-5 是早期代表；2007 年的三维 COF 工作进一步展示了通过构筑单元几何形状设计网络的思路。[早期二维 COF](https://doi.org/10.1126/science.1120411) · [三维 COF](https://doi.org/10.1126/science.1139915)

### 读图时关注什么？

- **构筑单元与连接键：** building block / linker 指参与构筑网络的分子单元；linkage 指连接这些单元的化学键或连接基团，两者不是同一个概念。
- **二维与三维网络：** 二维 COF 在层内形成共价网络，层间堆积影响孔道排列；三维 COF 的共价连接向三个空间方向延伸。图中的六方孔是一个例子，并不代表所有 COF 都是六方结构。
- **结构与功能：** 单元尺寸、连接几何、孔壁官能团和堆积方式共同影响孔道环境，因此要把孔径、组成和结构放在一起理解。
- **稳定性的适用条件：** “共价连接”不等于在任意条件下都稳定。连接化学、缺陷、溶剂、温度和酸碱条件都需要记录。β-酮烯胺 COF 的稳定性研究提供了一个具体案例，不能直接推广到所有 COF。[原始研究](https://doi.org/10.1021/ja308278w)

背景图中的 TpBD、TpAZ、TpDAAQ、TpBpy、TpPhen 用于引导比较不同构筑体系。图中平行 / 反平行及手性 / 非手性标注描述的是所示结构模型；判断具体材料的手性与堆积，需要结合三维结构和对称性，不能仅凭一个俯视图下结论。

### 为什么在 COF 课程中学习机器学习？

图中气体储存与分离、催化、能源与环境等应用，为提出材料问题提供了背景。本课程以 **CO₂ 吸附量预测** 为教学主线：将 COF 的结构和组成转成特征，用明确条件下的性质作为目标，再判断模型能否预测未见过的材料。

| 材料概念 | 数据中的表达 | 本课程的连接点 |
|---|---|---|
| 原子、晶胞与周期性 | CIF、晶格参数、元素与坐标 | 02：读取并检查结构 |
| 孔道与孔壁化学 | 孔径、孔隙率、密度、元素比例等 | 03：选择描述符 |
| 测量 / 模拟条件 | 温度、压力、单位、结构来源 | 04：审计真实数据 |
| 吸附性能 | 条件一致的吸附量与预测误差 | 05：训练与可靠验证 |

**思考：** 两个 COF 孔径相近，但孔壁官能团不同，它们的 CO₂ 吸附量一定相同吗？为什么训练数据要同时记录温度、压力和吸附量单位？

课程中的 `CO2_uptake_demo` 是人工构造的教学目标，用于学习流程，不能据此得出真实材料性能结论。

## 先看懂一张图

材料机器学习可以先理解成：

`材料样本 → 一组数字（feature） → 模型 → 预测值 → 与真实值比较`

例如，如果要预测 COF 的 CO₂ 吸附量：
- 材料样本：不同 COF；
- feature：孔径、孔隙率、密度、N/O 含量、官能团等；
- target：CO₂ 吸附量；
- model：学习 feature 与 target 之间关系的数学方法。

后续所有术语都围绕这条主线展开。

## 学习原则
1. 第一次看到新术语，不要求立即记住。
2. 先知道这个概念解决什么问题，再看公式。
3. 能运行代码只是第一步，必须知道输入是什么、输出是什么意思。
4. 先做简单 baseline，再考虑复杂模型。
5. 如果 01–05 能真正理解，这门入门课就已经达标。
