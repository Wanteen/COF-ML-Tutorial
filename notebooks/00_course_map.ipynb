## 先看懂一张图

材料机器学习可以先理解成：

`材料样本 → 一组数字（feature） → 模型 → 预测值 → 与真实值比较`

例如，如果要预测 COF 的 CO₂ 吸附量：
- 材料样本：不同 COF；
- feature：孔径、孔隙率、密度、N/O 含量、官能团等；
- target：CO₂ 吸附量；
- model：学习 feature 与 target 之间关系的数学方法。

后续所有术语都围绕这条主线展开。

## 学习原则
1. 第一次看到新术语，不要求立即记住。
2. 先知道这个概念解决什么问题，再看公式。
3. 能运行代码只是第一步，必须知道输入是什么、输出是什么意思。
4. 先做简单 baseline，再考虑复杂模型。
5. 如果 01–05 能真正理解，这门入门课就已经达标。
