# 00 课程地图：COF 机器学习从哪里开始？

本课程按“先成功一次，再逐步严谨”的节奏设计。

## 学习路线

`01 Python/data → 02 第一次回归+分类 → 03 CIF → 04 descriptors → 05 cleaning/feature engineering → 06 multiple models → 07 validation/tuning → 08 interpretation → 09 real COF case → 10 CIF-to-ML → 11 screening → 12 GNN → 13 MLFF`

### Stage 1 — 01–02 快速入门
先理解 X、y、fit、predict，并区分 regression 与 classification。

### Stage 2 — 03–08 核心方法
结构表示、特征构造、数据清洗、多模型、CV、调参和解释。

### Stage 3 — 09–13 COF 项目实践与进阶
真实 adsorption 数据、真实 CIF、高通量筛选、GNN、MLFF。

## 完成标准
完成 00–09 后，应能独立完成 `dataset → clean → features → regression/classification → model comparison → CV → final test → interpretation`。完成 10–11 后，应能解释 `CIF → descriptors → property mapping → screening → validation`。
