# 00 COF-ML-Tutorial 学习地图

本教程按“知识密度 + 难度”划分主章节；同一知识块中的短教程使用 A/B/C 编号。

### 掌握度标签
- 🟢 **Level A · 必须掌握**：完成 COF 机器学习基础训练所需。
- 🔵 **Level B · 建议掌握**：面向真实 COF 科研工作流与进阶表示。
- 🟣 **Level C · 了解即可**：拓展到机器学习势等专项主题。

## 推荐路线

| Chapter | Level | Topic | 学习目标 |
|---|---|---|---|
| 01A | 🟢 | Python & data basics | DataFrame、feature、target、X/y |
| 01B | 🟢 | First ML | 第一次回归 + 分类 |
| 02A | 🟢 | COF structure & CIF | 晶胞、PBC、pymatgen、结构 QC |
| 02B | 🟢 | COF descriptors | composition / crystal / pore descriptors |
| 02C | 🟢 | Data preparation | 清洗、特征构造与选择 |
| 03A | 🟢 | Model comparison | 多模型回归与分类比较 |
| 03B | 🟢 | Validation & tuning | CV、overfitting、leakage、调参 |
| 03C | 🟢 | Interpretation | correlation、permutation、SHAP |
| 04A | 🟢 | Real COF case study | 真实 CO₂ adsorption 综合训练 |
| 04B | 🔵 | CIF → ML | 真实 CIF 自动生成 ML table |
| 04C | 🔵 | Screening | 高通量预测、排序、适用域 |
| 05 | 🔵 | GNN | learned representation |
| 06 | 🟣 | MLFF | 原子能量/力与机器学习势 |

### 阶段性里程碑
**完成 01：** 我已经理解数据，并独立跑过回归和分类。  
**完成 02：** 我知道材料结构如何变成 ML-ready 数据。  
**完成 03：** 我会比较模型、验证结果并解释模型。  
**完成 04：** 我可以完成真实 COF ML 项目，并连接 CIF 与筛选。


## 每一块怎么学？
01：解释 X/y 并跑通回归/分类；02：说明特征来源并提交清洗记录；03：比较基线、隔离最终测试集、解释重要性的波动；04：提交数据卡、可追溯 ID 合并表和候选验证队列。05 建立图表示概念，06 是模拟方向的选读背景。练习用于自查，外部资源按具体任务扩展。
