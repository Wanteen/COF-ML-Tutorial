# COF × Machine Learning 入门路线

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/Wanteen/COF-ML-Tutorial/blob/main/notebooks/00_course_map.ipynb)

这套教程面向刚进入计算材料/COF 研究的研究生。目标不是快速调用一个黑箱模型，而是建立 **问题 → 数据 → 表示 → 模型 → 验证 → 科研解释** 的完整工作流。

## Learning map

| Stage | Question | Tools |
|---|---|---|
| 1 | 数据如何进入 Python？ | NumPy / pandas |
| 2 | 如何建立可靠 baseline？ | scikit-learn |
| 3 | CIF 如何变成结构对象？ | pymatgen |
| 4 | 结构如何变成 ML 特征？ | matminer + COF descriptors |
| 5 | 如何避免虚假的高精度？ | grouped/family-aware split |
| 6 | 能否让模型直接学习原子图？ | GNN / MatGL / ALIGNN |
| 7 | 能否学习能量和力？ | CHGNet / MACE / DeepMD |

推荐顺序：`00 → 01 → 02 → 03 → 04 → 05 → 06 → 07`。

## 完成后应该能够回答
- 一个 CIF 文件如何变成机器学习输入？
- composition descriptor、pore descriptor、graph representation 有什么区别？
- 为什么 random split 可能高估 COF 模型泛化能力？
- MAE、RMSE、R² 分别说明什么？
- 什么是 data leakage？
- 什么时候简单 baseline 比 GNN 更合理？
- 为什么孔道尺度和层间堆积对 COF 是特殊问题？
- MLFF 的 energy/force 标签来自哪里？
- 为什么预训练势“能运行”不等于“可用于你的 COF”？

## Course project
最终项目使用 `Wanteen/CURATED-COFs` 的真实结构/元数据。学生需要完成：数据审计 → 特征设计 → baseline → 严格划分 → 误差分析 → 科研结论。

**评价重点不是漂亮的 R²，而是能否证明模型学到了可泛化的材料规律。**
