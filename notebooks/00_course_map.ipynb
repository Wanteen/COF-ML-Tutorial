## COF 背景：从分子构筑到材料性质

**共价有机框架（Covalent Organic Frameworks, COFs）** 是有机构筑单元通过共价键连接形成的晶态多孔网络。理解 COF 时，需要同时考虑化学连接和周期结构。2005 年报道的 COF-1、COF-5 是早期代表；2007 年的三维 COF 工作进一步展示了通过构筑单元几何形状设计网络的思路。[早期二维 COF](https://doi.org/10.1126/science.1120411) · [三维 COF](https://doi.org/10.1126/science.1139915)

二维 COF 的层内共价网络、层间堆积、孔壁官能团和孔结构会共同影响材料性质。图中的六方孔只是一个例子，并不代表所有 COF。连接化学、缺陷、溶剂、温度和酸碱条件也都会影响稳定性。

本课程以 CO₂ 吸附/筛选问题作为背景，但会明确区分**教学 target**和**真实科研 target**。`CO2_uptake_demo` 是人工构造数据，只用于学习 ML 流程，不能据此得出材料性能结论。

## 课程真正要学会的主线

```text
真实 COF CIF
→ Structure / quality control
→ descriptors
→ COF_ID
→ target table
→ merge
→ train / test
→ model / validation
→ candidate screening
→ applicability domain
→ 回到 CIF / simulation / experiment 验证
```

早期版本的 02–05 已分别介绍 CIF、描述符、真实数据库和 baseline，但没有把它们真正串成一条从 CIF 到模型的闭环。05B 与 05C 专门补上这部分。

## 你需要具备什么基础？

**Python：** 能看懂变量、list/dict、`for` 循环、函数调用、`import` 和 DataFrame。

**COF / 材料基础：** 理解原子、化学键、晶胞、周期性结构、孔道、密度和吸附等基本概念。

**机器学习：不要求任何前置知识。**

## 课程分层

**Level A（必须掌握）：01–05C**

## 学习原则

1. 第一次看到新术语，不要求立即记住。
2. 先知道概念解决什么问题，再看公式。
3. 能运行代码只是第一步，必须知道输入是什么、输出是什么意思。
4. 先做可靠 baseline，再考虑复杂模型。
5. 不把人工 target 当成真实科研数据。
6. 不把最高 R² 当成唯一目标；要检查 split、leakage、适用域和 provenance。
7. **完成 Level A 的标准是理解 01–05C，而不是只会 `model.fit(X, y)`。**
