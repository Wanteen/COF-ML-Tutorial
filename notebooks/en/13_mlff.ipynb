# 13 Machine-Learned Force Fields (MLFF)

Ordinary supervised ML learns `descriptors → property`; MLFFs approximate `atomic configuration → energy / forces`.

Classical force fields are fast but constrained; DFT/AIMD is expensive; MLFFs learn energy/force relationships from quantum reference data. COFs may challenge transferability through stacking/sliding, vdW, flexibility, guests, water, ions and reactive configurations.

Research workflow: `DFT reference → train/fine-tune → independent test → exploratory MD → detect extrapolation → add data → retrain → production MD`.
