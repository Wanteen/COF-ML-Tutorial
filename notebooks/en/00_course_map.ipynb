# 00 Course Map

The course follows a ‘succeed once, then become rigorous’ progression.

`01 Python/data → 02 first regression & classification → 03 CIF → 04 descriptors → 05 cleaning/feature engineering → 06 multiple models → 07 validation/tuning → 08 interpretation → 09 real COF case → 10 CIF-to-ML → 11 screening → 12 GNN → 13 MLFF`

After 00–09 you should independently complete `dataset → clean → features → regression/classification → model comparison → CV → final test → interpretation`. After 10–11 you should also explain the full CIF-to-screening workflow.
