## COF background: from molecular building blocks to properties

**Covalent organic frameworks (COFs)** are crystalline porous networks assembled from organic building blocks through covalent bonds. Understanding a COF requires both chemical connectivity and periodic structure. COF-1 and COF-5 were early examples reported in 2005; 3D COF work in 2007 further demonstrated network design from building-block geometry. [Early 2D COFs](https://doi.org/10.1126/science.1120411) · [3D COFs](https://doi.org/10.1126/science.1139915)

Layer stacking, pore-wall functional groups, pore geometry and linkage chemistry can jointly affect properties. The hexagonal pores shown in the overview are examples rather than universal COF features. Stability also depends on defects, solvent, temperature and chemical environment.

The course uses CO₂ adsorption/screening as a scientific context but explicitly separates **teaching targets** from **research targets**. `CO2_uptake_demo` is synthetic and cannot support conclusions about real material performance.

## The workflow you should actually learn

```text
real COF CIF
→ Structure / quality control
→ descriptors
→ COF_ID
→ target table
→ merge
→ train / test
→ model / validation
→ candidate screening
→ applicability domain
→ return to CIF / simulation / experiment for verification
```

The earlier 02–05 chapters introduced CIF files, descriptors, real databases and baseline modeling separately. 05B and 05C now connect them into one complete workflow.

## What background do you need?

**Python:** basic variables, lists/dictionaries, loops, function calls, imports and DataFrames.

**COF / materials science:** atoms, bonds, unit cells, periodic structures, pores, density and adsorption.

**Machine learning: no prior knowledge is required.**

## Course levels

**Level A (core): 01–05C**
Start from basic data concepts and finish with real CIF → descriptors → target → validation → screening.

**Level B (recommended): 06 GNN**
Understand atomic graphs, message passing and learned representations.

**Supplementary: 07 MLFF**
Understand how ML interatomic potentials relate to DFT and MD; production training is not required.

## Learning principles

1. Do not memorize every term immediately; first understand what problem it solves.
2. Running code is only the first step; understand inputs and outputs.
3. Build a reliable baseline before trying a complex model.
4. Never treat a synthetic teaching target as scientific data.
5. Do not optimize only for the highest R²; check splitting, leakage, applicability domain and provenance.
6. **Level A is complete only when you understand 01–05C, not when you can merely call `model.fit(X, y)`.**
