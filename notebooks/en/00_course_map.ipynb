The course is grouped by knowledge density and difficulty. Short tutorials inside one knowledge block use A/B/C numbering.

## How to study each block
01: describe X/y and run regression/classification. 02: explain where features come from and submit a cleaning log. 03: compare baselines, isolate a final test and interpret uncertainty in importance. 04: submit a dataset card, traceable ID joins and a candidate validation queue. 05 develops graph concepts; 06 is optional simulation background. Exercises are checkpoints; data-source links are for task-specific extension.
