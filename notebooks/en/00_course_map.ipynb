This is not a specialist machine-learning course. The goal is to help new students in COF / computational materials research understand what machine learning is doing and complete a simple, reliable COF property-prediction workflow.

## What background do you need?

### Python
You only need to read basic code. You should recognize variables, lists/dictionaries, `for` loops, function calls, `import`, and DataFrames.

### COF / materials science
You should understand basic concepts such as atoms, chemical bonds, unit cells, periodic structures, pores and density.

### Machine learning
**No prior machine-learning knowledge is required.** We begin from the meaning of a row of data and a prediction target.

## Course levels

**Level A (core): 01–05**
Start from basic data concepts and finish with a COF baseline model.

**Level B (recommended): 06 GNN**
Understand atomic graphs and message passing. You do not need to implement a complex GNN from scratch.

**Supplementary: 07 MLFF**
Background only. Training or production use of ML potentials is not required.

## The big picture

A materials-ML workflow can first be understood as:

`material sample → numerical features → model → prediction → comparison with reference`

For example, if we want to predict CO₂ uptake in COFs:
- samples: different COFs;
- features: pore size, void fraction, density, N/O content, functional groups, etc.;
- target: CO₂ uptake;
- model: a mathematical method that learns a relationship between features and target.

Most terms in the course will connect back to this workflow.

## Learning principles
1. You do not need to memorize every new term immediately.
2. First understand what problem a concept solves, then look at formulas.
3. Running code is only the first step; know what the input and output mean.
4. Build a simple baseline before trying a complex model.
5. If you truly understand Chapters 01–05, you have completed the core course.
