## COF background: from molecular building blocks to properties

**Covalent organic frameworks (COFs)** are crystalline porous networks assembled from organic building blocks through covalent bonds. Understanding a COF requires both its chemical connectivity and its periodic arrangement in space. COF-1 and COF-5 were early examples reported in 2005; work on 3D COFs in 2007 extended the use of building-block geometry to design networks. [Early 2D COFs](https://doi.org/10.1126/science.1120411) · [3D COFs](https://doi.org/10.1126/science.1139915)

### What should you look for in the figure?

- **Building blocks and linkages:** a building block / linker is a molecular unit used to construct the network; a linkage is the chemical connection between units.
- **2D and 3D networks:** 2D COFs have covalent networks within layers, with stacking affecting pore alignment. In 3D COFs, covalent connectivity extends in three spatial directions. The illustrated hexagonal pore is one example, not a universal COF geometry.
- **Structure and function:** building-block size, connectivity, pore-wall groups and stacking jointly shape the pore environment. Pore size should therefore be interpreted together with chemistry and structure.
- **Stability under specified conditions:** covalent connectivity alone does not guarantee stability in every environment. Linkage chemistry, defects, solvent, temperature and acid/base conditions matter. A study of β-ketoenamine COFs provides a specific example, rather than evidence for all COFs. [Original study](https://doi.org/10.1021/ja308278w)

The TpBD, TpAZ, TpDAAQ, TpBpy and TpPhen labels invite comparison of different building-block systems. Parallel / antiparallel and chiral / achiral labels refer to the illustrated models. Assigning chirality or stacking to a material requires its 3D structure and symmetry, beyond a single top view.

### Why learn machine learning with COFs?

The figure introduces applications in gas storage and separation, catalysis, energy and the environment. This course uses **CO₂ uptake prediction** as a teaching example: represent COF structure and chemistry as features, define a target under specified conditions, and evaluate predictions on unseen materials.

| Materials concept | Data representation | Course connection |
|---|---|---|
| Atoms, unit cells and periodicity | CIF, lattice parameters, elements and coordinates | 02: inspect structures |
| Pores and pore-wall chemistry | Pore size, void fraction, density and elemental fractions | 03: choose descriptors |
| Measurement / simulation conditions | Temperature, pressure, units and structure provenance | 04: audit real data |
| Adsorption performance | Comparable uptake labels and prediction errors | 05: train and validate |

**Think about it:** if two COFs have similar pore sizes but different pore-wall groups, must they have the same CO₂ uptake? Why should a dataset record temperature, pressure and uptake units?

The course's `CO2_uptake_demo` is a synthetic teaching target for learning the workflow; it cannot support conclusions about real material performance.

## The big picture

A materials-ML workflow can first be understood as:

`material sample → numerical features → model → prediction → comparison with reference`

For example, if we want to predict CO₂ uptake in COFs:
- samples: different COFs;
- features: pore size, void fraction, density, N/O content, functional groups, etc.;
- target: CO₂ uptake;
- model: a mathematical method that learns a relationship between features and target.

Most terms in the course will connect back to this workflow.

## Learning principles
1. You do not need to memorize every new term immediately.
2. First understand what problem a concept solves, then look at formulas.
3. Running code is only the first step; know what the input and output mean.
4. Build a simple baseline before trying a complex model.
5. If you truly understand Chapters 01–05, you have completed the core course.
