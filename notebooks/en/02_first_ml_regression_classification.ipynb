# 02 Your First ML Models: Regression and Classification

Goal: **train a materials-ML model successfully and see a prediction.**

You will complete regression and classification before learning validation, tuning and interpretation.


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_absolute_error, r2_score, accuracy_score, ConfusionMatrixDisplay
df=pd.read_csv('https://raw.githubusercontent.com/Wanteen/COF-ML-Tutorial/main/data/cof_demo.csv')
display(df.head())
```

## 1. Regression

`X` contains material features and `y` is the target. `CO2_uptake_demo` is a synthetic teaching target only.


```python
features=['pore_A','void_fraction','density','N_fraction','O_fraction']
X=df[features]; y=df['CO2_uptake_demo']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
reg=RandomForestRegressor(n_estimators=200,random_state=42).fit(X_train,y_train)
pred=reg.predict(X_test)
print('MAE =',mean_absolute_error(y_test,pred),'R² =',r2_score(y_test,pred))
plt.scatter(y_test,pred); plt.xlabel('Actual'); plt.ylabel('Predicted'); plt.show()
```

The first complete loop is `features → split → fit → predict → metric`.


## 2. Classification

Use the median only as a teaching threshold: `1 = High`, `0 = Low`. Research thresholds require scientific justification.


```python
threshold=df['CO2_uptake_demo'].median()
yc=(df['CO2_uptake_demo']>=threshold).astype(int)
X_train,X_test,y_train,y_test=train_test_split(X,yc,test_size=0.25,random_state=42,stratify=yc)
clf=RandomForestClassifier(n_estimators=200,random_state=42).fit(X_train,y_train)
p=clf.predict(X_test)
print('accuracy =',accuracy_score(y_test,p))
ConfusionMatrixDisplay.from_predictions(y_test,p); plt.show()
```

## Regression vs classification
Regression predicts a continuous value; classification predicts a class. Typical metrics are MAE/RMSE/R² versus accuracy/precision/recall/F1.
