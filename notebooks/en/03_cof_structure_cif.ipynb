Chapter 02 used a ready-made table. This chapter asks where material features originate: unit cells, coordinates, periodicity and CIF parsing.


```python
!pip -q install pymatgen
from pymatgen.core import Lattice, Structure
lattice=Lattice.from_parameters(a=18,b=22,c=3.6,alpha=90,beta=90,gamma=95)
print(lattice); print('volume =',lattice.volume)
print(Lattice.hexagonal(a=12,c=3.5))
```

```python
s=Structure(lattice,['C','C','N','N'],[[0,0,0.5],[0.5,0.5,0.5],[0.25,0.25,0.5],[0.75,0.75,0.5]])
print(s.composition,len(s),s.density)
for site in s: print(site.species_string,site.frac_coords,site.coords)
```

## Real workflows start from CIF
Use `Structure.from_file("your_cof.cif")`. Check lattice parameters, composition, guests, missing H, disorder/occupancy and stacking before featurization.
