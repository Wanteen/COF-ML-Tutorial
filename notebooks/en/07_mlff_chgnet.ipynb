## 1. Three common approaches to atomistic simulation

### Classical force fields
Use predefined mathematical forms and parameters for atomic interactions. They are fast and can reach large systems and long MD times, but their accuracy and transferability are limited by the chosen functional form and parameterization.

### DFT / AIMD
Compute energies and forces from electronic structure. They are generally much more expensive. AIMD is especially costly because an electronic-structure calculation is performed at every MD step.

### MLFF
Train a machine-learning model on high-quality reference data so that it approximates the potential-energy surface and predicts energies and atomic forces at lower cost.

## 2. What is a potential-energy surface?
A particular arrangement of atoms corresponds to an energy. When atomic positions change, the energy changes.

A simplified mapping is: `atomic positions → energy`.

Atomic forces are related to how energy changes with coordinates. MLFF datasets therefore often contain:
- energy;
- force;
- sometimes stress / virial.

At this level, remember that an MLFF is not merely an ordinary property-regression model; it learns the energy/force relationships needed for atomistic motion.

## 3. Why are MLFFs attractive?
The ideal goal is to approach:
- DFT-level accuracy;
- classical-MD speed and scalability.

Whether this works depends strongly on whether the training data cover the structural configurations explored by the target simulation.

## 4. What is a pretrained MLFF?
Some models are trained on large materials datasets and can be loaded directly. This is similar to starting from a model that has already seen many structures.

However, 'the model runs' does not mean 'the model is reliable for your COF'. Check:
- whether relevant elements and chemical environments are represented;
- whether interlayer interactions, flexible structures, guests, water and ions are covered;
- whether energy/force errors are acceptable on your own DFT reference structures.

```python
# Optional API demonstration only; this is not a COF reliability test
!pip -q install chgnet pymatgen
from pymatgen.core import Lattice, Structure
from chgnet.model import CHGNet
structure=Structure(Lattice.cubic(4.2),['Li','Cl'],[[0,0,0],[0.5,0.5,0.5]])
model=CHGNet.load()
pred=model.predict_structure(structure)
print('Energy/atom:',pred['e'])
print('Forces shape:',pred['f'].shape)
```

## 5. Why COFs require particular caution
COFs may involve:
- interlayer vdW interactions, stacking and slip;
- framework flexibility;
- water and guest molecules;
- ions;
- high-energy or distorted configurations;
- in some systems, bond rearrangement or proton transfer.

If these situations are outside the training domain, a pretrained potential may extrapolate into unreliable regions.

## 6. Research-level workflow: background only
`DFT reference data → train/fine-tune MLFF → independent test → exploratory MD → detect extrapolation/problems → add reference data → retrain → production MD`

## Glossary
- force field: model of interatomic interactions;
- DFT: density functional theory;
- AIMD: ab initio molecular dynamics;
- potential energy surface: energy as a function of atomic configuration;
- MLFF: machine-learned force field / interatomic potential;
- pretrained model: model trained previously on a large dataset;
- fine-tuning: adapting a pretrained model to new data;
- active learning: iterative data selection and model improvement.
