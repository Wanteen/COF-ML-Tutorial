The goal of this chapter is not to train a model. It is to learn how to inspect a real dataset first. Research datasets often contain missing values, duplicates, inconsistent units and unclear field definitions.

```python
import pandas as pd
url='https://raw.githubusercontent.com/Wanteen/CURATED-COFs/master/cof-frameworks.csv'
df=pd.read_csv(url)
print('rows, columns =',df.shape)
display(df.head())
```

## 1. Ask two basic questions first
- What does each row represent? Usually one COF record.
- What does each column represent? It may be an ID, structural information, a property, a paper source or other metadata.

If these questions are unclear, you are not ready to train a model.

```python
print(df.columns.tolist())
summary=pd.DataFrame({
 'dtype':df.dtypes.astype(str),
 'missing':df.isna().sum(),
 'missing_%':(100*df.isna().mean()).round(1),
 'unique':df.nunique(dropna=True)
})
display(summary.head(30))
```

## 2. What is a missing value?
A missing value means that the database does not contain a valid value in that cell. It does **not** automatically mean zero.

For example, an unreported pore diameter does not mean a pore diameter of 0 Å. Filling every missing value with zero can introduce incorrect physical meaning.

```python
display(summary.sort_values('missing_%',ascending=False).head(15))
```

## 3. Metadata is not the same as a feature
A paper DOI, database ID or file name may be important for tracing provenance, but it is usually not a physical property of the material.

Keep useful metadata for traceability, but do not automatically use it as model input.

## 4. Why check duplicates?
If the same COF, or an almost identical COF, appears in both training and test sets, the test can become unrealistically easy. This is one possible source of **data leakage**.

## 5. Even the same property may not be directly comparable
CO₂ uptake depends on temperature, pressure and experimental or computational protocol. A target is therefore not merely 'a number'; you must confirm that values were generated under comparable conditions.

## Glossary
- missing value: unavailable or invalid entry;
- metadata: information describing identity or provenance;
- duplicate: repeated record;
- data leakage: inappropriate transfer of test information into training;
- provenance: where data came from and how it was generated.
