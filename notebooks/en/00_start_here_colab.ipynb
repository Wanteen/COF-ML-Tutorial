## How to use Colab
1. Run the initialization cell below.
2. The GitHub repository will be cloned to `/content/COF-ML-Tutorial`.
3. Later notebooks can read files from the repository `data/` folder.
4. Temporary files may disappear when the Colab session ends.
5. Editing a notebook in Colab does not automatically modify the instructor's GitHub repository.

```python
import os, subprocess
repo='/content/COF-ML-Tutorial'
if not os.path.exists(repo):
    subprocess.run(['git','clone','https://github.com/Wanteen/COF-ML-Tutorial.git',repo],check=True)
else:
    subprocess.run(['git','-C',repo,'pull'],check=True)
os.chdir(repo)
print('Working directory:',os.getcwd())
print('Repository ready.')
```
