## Run a lesson independently
Open the lesson in Colab, connect a runtime, then run cells from top to bottom. Each lesson imports its own dependencies; data-download cells require network access. A runtime reset clears Python variables and temporary files. Save exported tables before disconnecting. Heavy CHGNet installation belongs only to optional Level C.


## When a cell fails
`NameError`: run earlier cells first. HTTP errors: check the linked source and retry the download. Missing columns: inspect `df.columns` and confirm the dataset/version instead of renaming by guess. Package errors: restart after installation and record versions. Model scores vary with split and library version; explain changes rather than seeking one exact number.
