Machine-learning models do not directly understand statements such as 'this COF has a large pore' or 'this framework contains more nitrogen'. We need to convert materials information into numbers. Such numerical representations are called **descriptors** or features.

## COF context: turn a chemical question into features

The [background figure](https://raw.githubusercontent.com/Wanteen/COF-ML-Tutorial/main/assets/cof_background.jpg) compares building-block systems and pore structures. Changing a linker can affect both pore geometry and pore-wall chemistry. For CO₂ uptake, record complementary features such as pore size, void fraction, density and chemical composition; elemental fractions alone cannot identify topology or stacking.

Treat temperature and pressure consistently: either compare uptake at fixed conditions or include those conditions explicitly in the model inputs. Adsorption capacity, selectivity and transport are different targets. Choosing a target comes before choosing descriptors.

## 1. A simple example
For one COF we may record:
- density = 0.55 g/cm³;
- pore diameter = 22 Å;
- void fraction = 0.72;
- N fraction = 0.08.

These four numbers form a simple representation of the material.

```python
import pandas as pd
df=pd.DataFrame({
 'COF':['A','B','C'],
 'density':[0.55,0.78,0.61],
 'pore_A':[28.0,17.5,22.1],
 'void_fraction':[0.72,0.51,0.65],
 'N_fraction':[0.08,0.13,0.05]
})
df
```

## 2. What kinds of descriptors can a COF have?

### Composition / chemistry
Examples: C/N/O/F fractions, functional groups and linkage type.

### Geometry
Examples: pore diameter, void fraction, density and surface area.

### Structure / topology
Examples: topology, 2D/3D character, stacking and interlayer distance.

Different targets require different information. A feature set useful for adsorption does not have to be optimal for band-gap prediction.

## 3. Why is chemical composition alone not enough?
Two COFs may have similar elemental composition but very different pore sizes, topologies or stacking arrangements. Composition descriptors can therefore be useful, but they cannot capture every important COF structural feature.

```python
!pip -q install pymatgen matminer
from pymatgen.core import Composition
from matminer.featurizers.composition import ElementProperty
formulas=['C6H6','C6H4N2','C6H4O2']
tmp=pd.DataFrame({'composition':[Composition(x) for x in formulas]})
feat=ElementProperty.from_preset('magpie')
out=feat.featurize_dataframe(tmp,col_id='composition',ignore_errors=True)
print('Number of automatically generated columns:',out.shape[1]-1)
display(out.iloc[:,:8])
```

The matminer example only shows that composition descriptors can be generated automatically. **A large number of generated features does not mean they are all appropriate for your scientific question.**

For COFs, pore size, surface area, void fraction, topology, functional-group position and stacking often require additional calculations or data sources.

## Glossary
- representation: how a material is expressed to a model;
- descriptor: a numerical summary of materials information;
- composition descriptor: numerical information based on composition;
- geometric descriptor: numerical information based on geometry;
- categorical feature: a text category such as `linkage='imine'`.
