# 04 COF Descriptors: From Structure to ML Features

Materials information must be converted into numerical features or descriptors.


| Level | Examples | Source |
|---|---|---|
| Composition | C/N/O/F fraction | formula / CIF |
| Crystal | a,b,c,angles,volume,density | CIF |
| Pore | PLD,LCD,ASA,void fraction,pore volume | pore software |
| Chemistry | functional groups, local descriptors | structure tools |
| Learned | graph embedding | GNN |


```python
import pandas as pd
example=pd.DataFrame({'COF':['A','B','C'],'density':[0.55,0.78,0.61],'PLD_A':[12,7.5,10.1],'LCD_A':[18,11.2,15.4],'void_fraction':[0.72,0.51,0.65],'N_fraction':[0.08,0.13,0.05]})
example['LCD_PLD_ratio']=example['LCD_A']/example['PLD_A']
display(example)
```

Physically motivated combinations such as `LCD/PLD` are feature engineering. More features do not automatically mean a better model.
