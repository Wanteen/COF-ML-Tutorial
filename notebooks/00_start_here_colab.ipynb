## Colab 使用方法
1. 运行下面的初始化代码；
2. GitHub 仓库会复制到 `/content/COF-ML-Tutorial`；
3. 后续 Notebook 可以读取仓库中的 `data/`；
4. Colab 会话结束后，本地临时文件可能消失；
5. 你在 Colab 中修改内容，不会自动修改老师的 GitHub 仓库。

````python
import os, subprocess
repo='/content/COF-ML-Tutorial'
if not os.path.exists(repo):
    subprocess.run(['git','clone','https://github.com/Wanteen/COF-ML-Tutorial.git',repo],check=True)
else:
    subprocess.run(['git','-C',repo,'pull'],check=True)
os.chdir(repo)
print('Working directory:',os.getcwd())
print('Repository ready.')
````
