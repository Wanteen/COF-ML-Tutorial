## 怎样独立运行一节课？
在 Colab 打开目标笔记本，连接运行时，再从上到下运行单元格。每节自行导入依赖；下载数据需要网络。重启运行时会清空变量和临时文件，退出前保存导出的表。CHGNet 等较重依赖只在选读的 Level C 安装。


## 常见报错怎么处理？
`NameError`：先运行前面的单元格。HTTP 错误：打开数据来源，确认链接后重试。列不存在：先看 `df.columns`，确认数据表与版本，不要猜列名。依赖报错：安装后重启并记录版本。模型分数可能随划分和软件版本变化，重点解释变化，不是追求某个固定数值。
