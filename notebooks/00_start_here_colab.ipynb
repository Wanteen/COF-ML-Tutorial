## 使用方式
1. 在 Colab 中打开本 Notebook；
2. 运行下面的初始化代码；
3. 仓库会被克隆到 `/content/COF-ML-Tutorial`；
4. 后续 Notebook 可以统一使用仓库中的 `data/` 和其他文件；
5. 你在 Colab 中的修改默认只存在当前 Colab 会话，不会自动修改 GitHub 仓库。

````python
import os, subprocess, pathlib
repo = '/content/COF-ML-Tutorial'
if not os.path.exists(repo):
    subprocess.run(['git','clone','https://github.com/Wanteen/COF-ML-Tutorial.git', repo], check=True)
else:
    subprocess.run(['git','-C',repo,'pull'], check=True)
os.chdir(repo)
print('Working directory:', os.getcwd())
print('Repository ready.')
````

## Git / Colab 重要说明

`git clone` 只是在某一时刻复制仓库。老师之后更新 GitHub，学生本地或 Colab 中的旧副本 **不会自动更新**。

更新副本需要运行：

```bash
git pull
```

如果学生修改了文件但没有 `push`，老师的 GitHub 仓库也不会发生变化。
